```python
#1
import pandas as pd
import numpy as np
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/cvs files/Student Performance new.csv", index_col = 0)
df.head()
```

```python
dfr = df.rename(columns={"race/ethnicity": "race","parental level of education":"paredu","test preparation course":"prep","math percentage": "math","reading score percentage":"reading","writing score percentage":"writing"})
dfr
```

```python
#2
dfr.groupby("paredu").math.mean()
```

```python
#3
#Children of parents with master's degrees have the highest average score while children of parents with the high school graduate level education have the lowest average.
```

```python
#4
dfr.groupby("paredu").reading.median()
```

```python
dfr.groupby("paredu").writing.mean()
```

```python
#4 According to the mean and median scores of both reading and writing, the higher the parents' education level is, the higher the score,except for "some college" & "some high school".
#Yes, we can conclude that the educational background of the parents has, we can conclude that the educational background of the parents have the same influence on reading and writing scores.for "some college" & "some high school", we can conclude that the educational background of the parents have the same influence on reading and writing scores.
```

```python
#5
dfr.sort_values(by=["paredu","race"])
```

```python
#6
dfr.groupby(["paredu","race"])[["math","reading","writing"]].agg([len,min,max,np.mean,np.std])
```

```python
from scipy.stats import ttest_ind
```

```python
#7
reading_m = dfr[dfr.sex == "M"].reading
reading_f = dfr[dfr.sex == "F"].reading
ttest_ind(reading_m, reading_f)
```

```python
# p-value < 0.05, which means the gender is significant on the reading score.
```

```python
#8
math_BA = dfr[dfr.paredu == "bachelor's degree"].math
math_MA = dfr[dfr.paredu == "master's degree"].math
ttest_ind(math_BA, math_MA)
```

```python
#p-value is greater than 0.05, therefore the parents' education level is not significant on the math score.
#dummy variables will be nicer like MA,BA,HS use boolean values
```
